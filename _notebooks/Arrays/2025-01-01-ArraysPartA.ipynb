# In-Depth Explanation of the 2015 AP CSA FRQ - Question 4 (Part A)

This question asks you to design an **interface** called `NumberGroup` that represents a group of integers. The interface should define a **single method** named `contains`, which determines if a given integer is in the group.

---

## What is an Interface?

An **interface** in Java:
- Defines a contract for classes that implement it.
- Only includes **method signatures** (no method bodies).
- Allows different classes to have consistent behavior.

In this case, the interface `NumberGroup` should:
- Contain one method, `contains(int num)`.
- Be implemented by other classes that define how to determine if a number is in the group.

---

## Problem Requirements

The interface `NumberGroup` should:
1. Include exactly **one method**: `boolean contains(int num)`.
   - This method takes an integer `num` and returns `true` if `num` is in the group, `false` otherwise.
2. Be flexible enough for different types of groups (e.g., groups of consecutive numbers, specific sets of numbers, etc.).

---

## Implementation of the `NumberGroup` Interface

Below is how the `NumberGroup` interface should look:

```java
// Defining the NumberGroup interface
public interface NumberGroup {
    /**
     * Determines if a number is in the group.
     * @param num the number to check
     * @return true if num is in the group, false otherwise
     */
    boolean contains(int num);
}


````java
// Define the NumberGroup interface
public interface NumberGroup {
    boolean contains(int num);
}

// Define the Range class that implements NumberGroup
public class Range implements NumberGroup {
    private int lowerBound; // Starting value of the range
    private int upperBound; // Ending value of the range

    // Constructor to initialize the range
    public Range(int lower, int upper) {
        lowerBound = lower;
        upperBound = upper;
    }

    // Implements the contains method from the NumberGroup interface
    @Override
    public boolean contains(int num) {
        return num >= lowerBound && num <= upperBound;
    }
}

// Main class to test the Range class
public class Main {
    public static void main(String[] args) {
        // Create a range from 1 to 10
        Range range = new Range(1, 10);

        // Test the contains method
        System.out.println(range.contains(5));  // Output: true
        System.out.println(range.contains(11)); // Output: false
        System.out.println(range.contains(1));  // Output: true
    }
}
Main.main(null)
````

### Explanation of the Code

#### **`NumberGroup` Interface**
The `NumberGroup` interface is designed to represent a group of integers. It includes a single method, `contains(int num)`, which determines whether a given integer belongs to the group. This satisfies the prompt requirement to define an interface with exactly one method.

#### **`Range` Class**
The `Range` class implements the `NumberGroup` interface and represents a specific type of group—a continuous range of integers. Key details include:
- **Fields**: `lowerBound` and `upperBound` define the start and end of the range.
- **Constructor**: Initializes the range with the given bounds.
- **`contains` Method**:
  - Implements the method required by `NumberGroup`.
  - Checks whether the input integer lies between `lowerBound` and `upperBound` (inclusive).
  - Returns `true` if the integer is within the range, otherwise `false`.

#### **`Main` Class**
The `Main` class is used to test the `Range` implementation. It creates a `Range` object and tests the `contains` method with various inputs:
- Example: A range from 1 to 10.
  - `contains(5)` → `true` (5 is within the range).
  - `contains(11)` → `false` (11 is outside the range).
  - `contains(1)` → `true` (1 is the lower bound of the range).

#### **How It Satisfies the Prompt**
1. **Interface Requirement**: The `NumberGroup` interface defines exactly one method, `contains`, as specified.
2. **Concrete Implementation**: The `Range` class provides an example of how to represent a group of integers using the interface.
3. **Demonstration**: The `Main` class verifies the correctness of the implementation with practical examples, ensuring the functionality aligns with the prompt's description.
